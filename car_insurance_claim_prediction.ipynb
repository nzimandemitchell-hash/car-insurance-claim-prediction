```python
# Fill missing credit scores with the median
df["credit_score"] = df["credit_score"].fillna(
    df["credit_score"].median()
)

# Fill missing annual mileage with the median
df["annual_mileage"] = df["annual_mileage"].fillna(
    df["annual_mileage"].median()
)

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
print("Dataset shape:", df.shape)

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate IDs:")
print(df["id"].duplicated().sum())
```

```python
df_clean = df.copy()

print("Clean dataset created successfully!")
print(df_clean.shape)
```

```python
df_clean.info()
```

```python
# Separate features from target

X = df_clean.drop(columns=["outcome", "id"])
y = df_clean["outcome"]

print("Features (X):", X.shape)
print("Target (y):", y.shape)
```

```python
categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_columns = X.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("\nNumerical columns:")
print(numerical_columns)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        ),
        (
            "numerical",
            "passthrough",
            numerical_columns
        )
    ]
)

print("Preprocessor created successfully!")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Training claim rate:")
print(y_train.mean() * 100)

print("\nTesting claim rate:")
print(y_test.mean() * 100)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_columns
        )
    ]
)
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000))
    ]
)

print("Model pipeline created successfully!")
```

```python
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Predictions generated!")
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2%}")
```
